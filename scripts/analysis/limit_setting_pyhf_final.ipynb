```python
multisim_uncertainty = 0.2   # 20% flux/xsec
detvars_uncertainty = 0.3    # 30% detector variations
dirt_uncertainty = 0.5       # 50% dirt normalization

if(not tests):

    delimiter = 0
    pot_uncert = 0.02  # 2% POT uncertainty 

    # Use our histogram arrays instead of ROOT files
    total_bkg = np.sum(hist_bkg_run1) + np.sum(hist_bkg_run3)
    total_run1 = np.sum(hist_bkg_run1)
    total_run3 = np.sum(hist_bkg_run3)

    # Combine Run1 and Run3 backgrounds
    n_back = hist_bkg_run1.tolist()[delimiter:]
    n_back.extend(hist_bkg_run3.tolist()[delimiter:])

    # Combine Run1 and Run3 data
    n_data = hist_data_run1.tolist()[delimiter:]
    n_data.extend(hist_data_run3.tolist()[delimiter:])

    # Background uncertainties
    if(flat_uncertainty):
        back_sigma = (hist_bkg_run1 * uncertainty).tolist()[delimiter:]
        back_sigma.extend((hist_bkg_run3 * uncertainty).tolist()[delimiter:])
    else:
        # Calculate uncertainties from components (stat + syst)
        # For nu: add multisim (20%) and detvars (30%) in quadrature
        # For dirt: add 50% uncertainty
        # For offbeam: statistical only
        
        sigma_nu_run1 = np.sqrt((hist_nu_run1 * multisim_uncertainty)**2 + (hist_nu_run1 * detvars_uncertainty)**2)
        sigma_dirt_run1 = hist_dirt_run1 * dirt_uncertainty
        sigma_offbeam_run1 = np.sqrt(hist_offbeam_run1 * offbeam_scaling_run1)  # Poisson
        sigma_bkg_run1 = np.sqrt(sigma_nu_run1**2 + sigma_dirt_run1**2 + sigma_offbeam_run1**2)
        
        sigma_nu_run3 = np.sqrt((hist_nu_run3 * multisim_uncertainty)**2 + (hist_nu_run3 * detvars_uncertainty)**2)
        sigma_dirt_run3 = hist_dirt_run3 * dirt_uncertainty
        sigma_offbeam_run3 = np.sqrt(hist_offbeam_run3 * offbeam_scaling_run3)  # Poisson
        sigma_bkg_run3 = np.sqrt(sigma_nu_run3**2 + sigma_dirt_run3**2 + sigma_offbeam_run3**2)
        
        back_sigma = sigma_bkg_run1.tolist()[delimiter:]
        back_sigma.extend(sigma_bkg_run3.tolist()[delimiter:])

    if alpha == 1.0:
        scaling_a1 = (1./pow(0.1, 3))
    else:
        scaling_a1 = 1. 

    for mass in masses:

        print("Processing mass: " + mass)
        
        # Use our histogram arrays
        total_sig_run1 = np.sum(hist_signal_run1) * scaling_a1
        total_sig_run3 = np.sum(hist_signal_run3) * scaling_a1
        
        print("Signal events run1: {nevts:.5f}".format(nevts=total_sig_run1))
        print("Signal events run3: {nevts:.5f}".format(nevts=total_sig_run3))

        total_sig = total_sig_run1 + total_sig_run3
        
        # Calculate scaling factor to get mu ~ 1
        factor = scaling * (((total_run1/total_sig_run1) + (total_run3/total_sig_run3)) / 2.)
        scaling_list.append(factor)
        
        print("Scaling factor: ", factor)

        # Apply scaling and fraction_outside correction to signal
        n_sig_run1 = (hist_signal_run1 * scaling_a1 * fraction_outside * factor).tolist()[delimiter:]
        n_sig_run3 = (hist_signal_run3 * scaling_a1 * fraction_outside * factor).tolist()[delimiter:]
        n_sig_run1.extend(n_sig_run3)

        # Signal uncertainties (use flat 20% for now since we don't have per-bin errors)
        signal_uncert = 0.20  # 20% signal uncertainty
        sigma_sig_run1 = (hist_signal_run1 * scaling_a1 * fraction_outside * factor * signal_uncert).tolist()[delimiter:]
        sigma_sig_run3 = (hist_signal_run3 * scaling_a1 * fraction_outside * factor * signal_uncert).tolist()[delimiter:]
        sigma_sig_run1.extend(sigma_sig_run3)

        model = pyhf.Model(
            {
          "channels": [
            {
              "name": "singlechannel",
              "samples": [
                {
                  "name": "signal",
                  "data": n_sig_run1,
                  "modifiers": [
                    {"name": "mu", "type": "normfactor", "data": None},
                    {"name": "uncorr_siguncrt", "type": "shapesys", "data": sigma_sig_run1},
                    {"name": "pot_correlated", "type": "normsys", "data": {"hi":1. + (pot_uncert), "lo": 1. - (pot_uncert)}},
                    {"name": "flux_correlated", "type": "normsys", "data": {"hi":1. + flux_uncert, "lo": 1. - flux_uncert}},
                  ]
                },
                {
                  "name": "background",
                  "data": n_back,
                  "modifiers": [
                    {"name": "uncorr_bkguncrt", "type": "shapesys", "data": back_sigma},
                    {"name": "pot_correlated", "type": "normsys", "data": {"hi":1. + pot_uncert, "lo": 1 - pot_uncert}},
                  ]
                }
              ]
            }
          ]
        }
        )

        obs = n_data + model.config.auxdata
        poi_values = np.linspace(0., 10., 100)
        obs_limit, exp_limits, (scan, results) = pyhf.infer.intervals.upperlimit(
            obs, model, poi_values, level=0.1, return_results=True)
        
        CLs_value, p_values, CLs_band = pyhf.infer.hypotest(obs_limit, obs, model, 
                                                           return_expected_set=True, 
                                                           return_tail_probs=True)
        
        print(f"Upper limit (obs): μ = {obs_limit:.4f}")
        print(f"Upper limit (exp): μ = {exp_limits[2]:.4f}")
        print(f"p-value: = {p_values[1]:.2f}")

        obs_epsilon = (nominal_eps**2)*np.sqrt(obs_limit*factor)
        exp_epsilon = (nominal_eps**2)*np.sqrt(exp_limits[2]*factor)
        exp_two_down = (nominal_eps**2)*np.sqrt(exp_limits[0]*factor)
        exp_one_down = (nominal_eps**2)*np.sqrt(exp_limits[1]*factor)
        exp_one_up = (nominal_eps**2)*np.sqrt(exp_limits[3]*factor)
        exp_two_up = (nominal_eps**2)*np.sqrt(exp_limits[4]*factor)

        obs_eps.append(obs_epsilon)
        exp_eps.append(exp_epsilon)
        exp_minus_two.append(exp_two_down)
        exp_minus_one.append(exp_one_down)
        exp_plus_one.append(exp_one_up)
        exp_plus_two.append(exp_two_up)
        CLb_values.append(p_values[1])
        
        print(f"Upper limit (obs): epsilon2 = {obs_epsilon}")
        print(f"Upper limit (exp): epsilon2 = {exp_epsilon}")
        print(f"Upper limit +2sigma: epsilon2 = {exp_two_up}")
        print(f"Upper limit +1sigma: epsilon2 = {exp_one_up}")
        print(f"Upper limit -1sigma: epsilon2 = {exp_one_down}")
        print(f"Upper limit -2sigma: epsilon2 = {exp_two_down}")
        print("\n")
```
